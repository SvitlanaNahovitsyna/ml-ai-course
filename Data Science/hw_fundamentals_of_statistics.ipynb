**Завдання 1.** Ми обговорили з Вами на лекції, що велика літера Сігма грецького алфавіту позначає в математиці суму. Знайдіть з використанням Python $\Sigma x$, $\Sigma x^2$, та $\Sigma (x-1)^2$ для наступного набору даних:
`1 3 4`
Виведіть результат кожного обчислення на екран.

```python
import numpy as np
```

```python
# Pure Python
x = [1, 3, 4]
sum_x = sum(x)
sum_x2 = sum(x**2 for x in x)
sum_x_minus_1_sq = sum((x-1)**2 for x in x)

print("Pure Python:")
print(f"Σx = {sum_x}")
print(f"Σx² = {sum_x2}")
print(f"Σ(x - 1)² = {sum_x_minus_1_sq}")
print("-" * 20)

# NumPy
x_np = np.array([1, 3, 4])
sum_x_np = np.sum(x_np)
sum_x2_np = np.sum(x_np**2)
sum_x_minus_1_sq_np = np.sum((x_np-1)**2)

print("Python with NumPy:")
print(f"Σx = {sum_x_np}")
print(f"Σx² = {sum_x2_np}")
print(f"Σ(x - 1)² = {sum_x_minus_1_sq_np}")
```

**Завдання 3.** Для того ж набору даних в змінній `sum_purchases` обчисліть **медіану** кожним з зазначених методів:  

  3.1. без використання жодних бібліотек, користуючись тільки Python.  
  3.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

```python
sorted_purchases = sorted(sum_purchases)
n = len(sum_purchases)

if n % 2 != 0:
    median_pure = sorted_purchases[n // 2]
else:
    median_pure = (sorted_purchases[n // 2 - 1] + sorted_purchases[n // 2]) / 2

print(f"Медіана (Pure Python): {median_pure}")

median_np = np.median(sum_purchases)
print("Медіана (NumPy):", median_np)

print("Чи співпадають результати?", median_pure == median_np)
```

Наше вибіркове середнє дорівнює 667.95, а медіана - 319.0.
Вибіркове середнє значення набагото вище за медіану. За рахунок того, що э кылька покупців з великими чеками, вибіркове середнє стає більшим за рахунок цього. Тоді як медіана (319.0) залишається стійкою до таких «викидів» і показує більш реальну суму.

**Вибіркове середнє** - краще брати для розрахунку загальних обсягів (наприклад, загальної виручки магазину за місяць).

**Медіана** - це надійний показник для «неідеальних» даних із викидами.(наприклад, для оцінки реальної, типової покупки більшості клієнтів — 319).
Оскільки середня вартість спотворюється через екстремальні значення, саме медіана є найкращою мірою для асиметричних даних і допомагає точно оцінити типову покупку більшості клієнтів.

**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:  

  4.1. без використання жодних бібліотек, користуючись тільки Python.  
  4.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран.

```python
# 4.1. Без використання жодних бібліотек, користуючись тільки Python
n = len(sum_purchases)
mean_x = sum(sum_purchases) / n

# Сума квадратів відхилень
sum_squared_diff = sum([(el - mean_x) ** 2 for el in sum_purchases])

# Розрахунок стандартного відхилення
std_pure = (sum_squared_diff / (n - 1)) ** 0.5

# # Або
# import math
# std_pure = math.sqrt(sum_squared_diff / (n - 1))

print(f"Стандартне відхилення (Pure Python): {std_pure}")

# 4.2. За допомогою NumPy
std_np = np.std(sum_purchases, ddof=1)
print(f"Стандартне відхилення (NumPy): {std_np}")
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
# n та mean_x вже були знайдені у попередніх завданнях:
# n = len(sum_purchases) однаково з len(visits)
# mean_x = sum(sum_purchases) / n

# 1. Знаходимо середнє для visits
mean_y = sum(visits) / n

# 6.1. Коефіцієнт Пірсона (Pure Python через генератори)
num = sum((sum_purchases[i] - mean_x) * (visits[i] - mean_y) for i in range(n))
sum_sq_x = sum((sum_purchases[i] - mean_x) ** 2 for i in range(n))
sum_sq_y = sum((visits[i] - mean_y) ** 2 for i in range(n))

r_pure = num / ((sum_sq_x * sum_sq_y) ** 0.5)
print(f"Коефіцієнт Пірсона (Pure Python): {r_pure:}")

# 6.2. За допомогою бібліотеки NumPy
r_np = np.corrcoef(sum_purchases, visits)[0, 1]
print(f"Коефіцієнт Пірсона (NumPy): {r_np:}")
```

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

Аналіз зв'язку: Розрахунок коефіцієнта Пірсона обома методами дав однаковий результат.

**Аналіз за графіком розсіювання:** Графік показує наявність окремих точок-викидів, які сильно віддалені від основного масиву даних. Оскільки коефіцієнт кореляції Пірсона є дуже чутливим до викидів та нелінійних ефектів, ці поодинокі значення можуть спотворювати реальну картину.

**Що варто змінити:** необхідно попередньо очистити вибірку від викидів (аутльєрів) або застосувати до даних логарифмічне перетворення, щоб згладити екстремальні значення і наблизити зв'язок до лінійного.

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
std_x = np.std(sum_purchases, ddof=1)
threshold = mean_x + 2 * std_x
print(f"Поріг для визначення викидів: {threshold:}")

# 2. Фільтруємо обидва масиви синхронно
filtered_purchases = [sum_purchases[i] for i in range(n) if sum_purchases[i] <= threshold]
filtered_visits = [visits[i] for i in range(n) if sum_purchases[i] <= threshold]

# 3. Обчислюємо новий коефіцієнт Пірсона
r_cleaned = np.corrcoef(filtered_purchases, filtered_visits)[0, 1]
print(f"Коефіцієнт Пірсона після видалення викидів (NumPy): {r_cleaned:}")
```

```python
# Побудова діаграми розсіювання для очищених даних
plt.scatter(filtered_purchases, filtered_visits)
plt.title('Scatter Plot of Purchases vs. Visits (Filtered)')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

Після видалення викидів ми бачимо реальний розподіл даних на графіку. Коефіцієнт Пірсона став трохи більше мінусовим ~0.37. Це вказує на відсутність помітного лінійного зв'язку між сумою покупок і кількістю візитів. Точки на діаграмі розсіювання розташовані хаотично, а це означає, що частота відвідувань магазину практично не залежить від суми чека покупця.
